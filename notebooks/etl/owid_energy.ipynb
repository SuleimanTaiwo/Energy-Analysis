```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


```

```python
from sqlalchemy import create_engine
#Imports the function to load environment variables from an `.env` file into the Python environment
from dotenv import load_dotenv

# Imports the built-in `os` module to access those environment variables using os.getenv()
import os

load_dotenv(dotenv_path="C:/Users/Admin/project-data/Non_elec_energy_analysis/Energy-Analysis/.env")

# Retrieves the PostgreSQL credentials from the `.env` file environment variable 
username = os.getenv("DB_USER")
password = os.getenv("DB_PASSWORD")
host     = os.getenv("DB_HOST")
port     = os.getenv("DB_PORT")
database = os.getenv("DB_NAME")

# Create the SQLAlchemy engine
engine = create_engine(f"postgresql+psycopg2://{username}:{password}@{host}:{port}/{database}")

# check if the connection works
with engine.connect() as connection:
    print("SQLAlchemy connected!")
```

```python
owid_raw = pd.read_csv(r"C:\Users\Admin\project-data\Non_elec_energy_analysis\Energy-Analysis\data\owid-energy-data.csv")
```

```python
pd.set_option('display.max_info_columns', 200)
```

```python
owid_raw.info()
```

```python
countries = ["Nigeria", "South Africa", "Brazil", "Indonesia", "Mexico"]
```

```python
owid = owid_raw[owid_raw["country"].isin(countries)]
```

```python
owid = owid.reset_index(drop=True)
```

```python
owid.info()
```

```python
owid.isna().sum().sort_values()
```

```python
num_cols = owid.select_dtypes(include=['float64', 'int64']).columns
```

```python
owid[num_cols] = owid[num_cols].fillna(0)
```

```python
owid.isna().sum().sort_values()
```

```python
columns_needed = [
    "country",
    "iso_code",
    "year",
    "population",
    "gdp",

    "primary_energy_consumption",
    "energy_per_capita",
    "energy_per_gdp",

    "oil_consumption",
    "gas_consumption",
    "coal_consumption",
    "biofuel_consumption",
    "renewables_consumption",
    "low_carbon_consumption",
    "fossil_fuel_consumption",
    "fossil_share_energy",
    "low_carbon_share_energy",
    
    "greenhouse_gas_emissions"
]
```

```python
owid = owid[columns_needed]
```

```python
owid
```

```python
owid = owid.rename(columns={
      "primary_energy_consumption": "total_energy_twh",
    "energy_per_capita": "energy_per_capita_kwh",
    "energy_per_gdp": "energy_intensity",
    "fossil_fuel_consumption": "fossil_consumption_twh",
    "low_carbon_consumption": "low_carbon_consumption_twh",
    "renewables_consumption": "renewables_consumption_twh",
    "oil_consumption": "oil_consumption_twh",
    "gas_consumption": "gas_consumption_twh",
    "coal_consumption": "coal_consumption_twh",
    "biofuel_consumption": "biofuel_consumption_twh",
    "greenhouse_gas_emissions": "co2_emissions"
})
```

```python
owid
```

```python
owid.to_sql("energy_master",
           engine,
           if_exists="replace",
           index=False
           )
print("energy data successfully loaded into PostgreSQL.")
```
